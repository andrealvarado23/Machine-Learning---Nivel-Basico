```python
# ARANGE

print(np.arange(100)) # fin = 100

print(np.arange(10, 21)) # inicio = 10, fin = 21

print(np.arange(1, 100, 3)) # inicio = 1, fin = 100, paso = 3
```

```python
ADN = "CGTACGATCGATCGACTC"

for n in np.arange(0, len(ADN), 3): # 0, 3, 6, 9, ..., 15
  print(ADN[n:n+3])
```

```python
### LINSPACE

print(np.linspace(0, 10, 101)) # inicio = 1, fin = 10 (I), Cantidad = 10
```

```python
# Modelos del crecimiento bacteriano

N0 = 10 # Población inicial
r = 0.5 # Tasa de crecimiento
K = 10000 # Capacidad de carga

t = np.linspace(1, 30, 1000) # Serie de tiempo
N = K / (1 + ((K - N0) / N0) * np.exp(-r*t)) # Función logística

plt.plot(t, N)
```

```python
# MESHGRID

x = np.linspace(-5, 5, 11)
y = np.linspace(-5, 5, 11)

# X =       -5,   -4, -3, ...
# Y = -5 (-5, -5), (-5, -4)
#.    -4 (-4, -5)
#.    -3

_x, _y = np.meshgrid(x, y)

plt.plot(_x, _y, "o")
```

```python
x = np.linspace(-5, 5, 11)
y = np.linspace(-5, 5, 11)

_x, _y = np.meshgrid(x, y)

z = -np.sqrt(_x**2 + _y**2)

plt.matshow(z)
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

_x, _y = np.meshgrid(x, y)

z = -np.sqrt(_x**2 + _y**2)

fig = plt.figure()

ax = fig.add_subplot(111, projection = "3d")

ax.plot_surface(_x, _y, z, cmap = "viridis")

plt.show()
```

```python
### NUMEROS ALEATORIOS
rng = np.random.default_rng()
rng

# entre 0 y 1
print(rng.random(5))
print(rng.random((5, 6)))

# Enteros
print(rng.integers(1, 10, size = 5))
print(rng.integers(1, 10, size = (5, 4)))

# Normales
print(rng.normal(0, 1, 100))
print(rng.normal(10, 10, 100))
```

```python
#.             media, sd
datos = rng.normal(0, 1, 10000)
plt.hist(datos, bins = 30)
plt.show()
```
